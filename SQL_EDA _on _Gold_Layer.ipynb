# SQL EDA on Gold Layer

### 1. Top occupations in Karnataka

```python
%sql
SELECT
    o.NCO_Name,
    SUM(f.Total_Persons) AS workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_occupation o
ON f.Occupation_Key = o.Occupation_Key
GROUP BY o.NCO_Name
ORDER BY workers DESC
LIMIT 20;
```

### 2. Top districts by workforce

```python
%sql
SELECT
    g.District_Name,
    SUM(f.Total_Persons) AS workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_geography g
ON f.Geo_Key = g.Geo_Key
WHERE g.Geo_Level = 'DISTRICT_DETAIL'
GROUP BY g.District_Name
ORDER BY workers DESC;
```

### 3. SC vs ST workforce comparison

```python
%sql
SELECT
    c.Community_Category,
    SUM(f.Total_Persons) AS workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_community c
ON f.Community_Key = c.Community_Key
GROUP BY c.Community_Category;
```

### 4. Female workforce participation by district

```python
%sql
SELECT
    g.District_Name,
    SUM(f.Total_Females) AS female_workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_geography g
ON f.Geo_Key = g.Geo_Key
WHERE g.Geo_Level = 'DISTRICT_DETAIL'
GROUP BY g.District_Name
ORDER BY female_workers DESC;
```

### 5. Rural vs Urban workforce distribution

```python
%sql
SELECT
    SUM(Rural_Persons) AS rural_workers,
    SUM(Urban_Persons) AS urban_workers
FROM pyspark_dbt.gold.fact_occupation_workforce;
```

### 6. Occupations with highest female participation

```python
%sql
SELECT
    o.NCO_Name,
    SUM(f.Total_Females) AS female_workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_occupation o
ON f.Occupation_Key = o.Occupation_Key
GROUP BY o.NCO_Name
ORDER BY female_workers DESC
LIMIT 20;
```

### 7. Most dominant industries

```python
%sql
SELECT
    i.Industry_Name,
    SUM(f.Persons) AS workers
FROM pyspark_dbt.gold.fact_industrial_workforce f
JOIN pyspark_dbt.gold.dim_industry i
ON f.Industry_Key = i.Industry_Key
GROUP BY i.Industry_Name
ORDER BY workers DESC;
```

### 8. Main vs Marginal workers

```python
%sql
SELECT
    w.Worker_Type,
    SUM(f.Persons) AS workers
FROM pyspark_dbt.gold.fact_industrial_workforce f
JOIN pyspark_dbt.gold.dim_worker_type w
ON f.Worker_Type_Key = w.Worker_Type_Key
GROUP BY w.Worker_Type;
```

### 9. Top industries in Bengaluru district

```python
%sql
SELECT
    i.Industry_Name,
    SUM(f.Persons) AS workers
FROM pyspark_dbt.gold.fact_industrial_workforce f
JOIN pyspark_dbt.gold.dim_industry i
ON f.Industry_Key = i.Industry_Key
JOIN pyspark_dbt.gold.dim_geography g
ON f.Geo_Key = g.Geo_Key
WHERE g.District_Name = 'Bangalore'
GROUP BY i.Industry_Name
ORDER BY workers DESC;
```

### 10. Industry specialization by district

```python
%sql
SELECT
    g.District_Name,
    i.Industry_Name,
    SUM(f.Persons) AS workers
FROM pyspark_dbt.gold.fact_industrial_workforce f
JOIN pyspark_dbt.gold.dim_geography g
ON f.Geo_Key = g.Geo_Key
JOIN pyspark_dbt.gold.dim_industry i
ON f.Industry_Key = i.Industry_Key
WHERE g.Geo_Level = 'DISTRICT_DETAIL'
GROUP BY
    g.District_Name,
    i.Industry_Name
ORDER BY workers DESC;
```

### 11. Top occupations among SC workers

```python
%sql
SELECT
    o.NCO_Name,
    SUM(f.Total_Persons) AS workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_occupation o
ON f.Occupation_Key = o.Occupation_Key
JOIN pyspark_dbt.gold.dim_community c
ON f.Community_Key = c.Community_Key
WHERE c.Community_Category = 'Scheduled Caste (SC)'
GROUP BY o.NCO_Name
ORDER BY workers DESC
LIMIT 20;
```

### 12. Top occupations among ST workers

```python
%sql
SELECT
    o.NCO_Name,
    SUM(f.Total_Persons) AS workers
FROM pyspark_dbt.gold.fact_occupation_workforce f
JOIN pyspark_dbt.gold.dim_occupation o
ON f.Occupation_Key = o.Occupation_Key
JOIN pyspark_dbt.gold.dim_community c
ON f.Community_Key = c.Community_Key
WHERE c.Community_Category = 'Scheduled Tribe (ST)'
GROUP BY o.NCO_Name
ORDER BY workers DESC
LIMIT 20;
```
